```python
@node("src.mini", "Source")
def mini() -> pl.LazyFrame:
    return pl.LazyFrame({"x": [1, 2, 3, 4]})
```

```python
g = Graph()
g.add("s", "src.mini")
g.add("cx", "pl.col", {"name": "x"})
g.add("sum", "Expr.sum", inputs={"self": [("cx", "out")]})
g.add("sel", "LazyFrame.select", inputs={"self": [("s", "out")], "exprs": [("sum", "out")]})
g.add("out", "LazyFrame.collect", inputs={"self": [("sel", "out")]})
```
