# Plotting nodes (behaviz + bokeh)

End-to-end test of the `sink.plot.*` nodes: a graph -> an interactive bokeh plot rendered inline.
The node returns `{"type":"html", "html": ...}` with BokehJS **inlined** (self-contained, works offline).
`overrides` is a dict input that forwards **any** bokeh styling kwarg -- nothing is hidden.

```python
from IPython.display import HTML

import nota.nodes  # register builtin nodes (sources, sinks, plots)
from nota.core import Graph, run
```

## Line plot -- sample data
`sink.plot.line` with `x`/`y` column names.

```python
g = Graph()
g.add('src', 'source.sample')
g.add('plot', 'sink.plot.line', {'x': 'id', 'y': 'score', 'title': 'Score by id'})
g.connect('src', 'plot', 'frame')
out = run(g)['plot']
print(out['type'], '| html bytes:', len(out['html']))
HTML(out['html'])
```

## Styling via `overrides` (the un-hidden kwargs)
Any bokeh glyph kwarg passes straight through -- here as a literal dict param.
In the GUI this is the `overrides` input port (wire a `dict.build` node into it).

```python
g = Graph()
g.add('src', 'source.sample')
g.add('plot', 'sink.plot.scatter',
      {'x': 'id', 'y': 'score', 'title': 'Styled scatter',
       'overrides': {'size': 14, 'fill_color': 'crimson', 'fill_alpha': 0.6}})
g.connect('src', 'plot', 'frame')
HTML(run(g)['plot']['html'])
```

## Bar + step
Same pattern, different mark.

```python
for kind, title in [('sink.plot.bar', 'Bar'), ('sink.plot.step', 'Step')]:
    g = Graph()
    g.add('src', 'source.sample')
    g.add('plot', kind, {'x': 'id', 'y': 'score', 'title': title})
    g.connect('src', 'plot', 'frame')
    display(HTML(run(g)['plot']['html']))
```
