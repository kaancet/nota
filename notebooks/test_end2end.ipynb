### Single filter

```python
g = Graph()
g.add("node_source","source.csv",params={"path":"test.csv"})

# filtering
g.add("cv",   "pl.col",     params={"name": "id"}) # column id
g.add("pred", "Expr.gt",    params={"other": 4}) # greater than

g.add("node_filter","LazyFrame.filter") #  <- this is where filtering node comes
g.add("node_preview","sink.preview")

g.connect("cv","pred")
g.connect("node_source","node_filter",port="self")

g.connect("pred","node_filter",port="predicates")
g.connect("node_filter","node_preview")
```

```python
draw(g)
```

```python
run(g)
```

### Multi filter

```python
g = Graph()
# source
g.add("node_source","source.csv",params={"path":"test.csv"})

# filter 1
g.add("c_id",   "pl.col",     params={"name": "id"}) # column id
g.add("pred_gt", "Expr.gt",    params={"other": 4}) # greater than

# filter 2
g.add("c_val","pl.col",  params={"name": "value"}) # column value
g.add("pred_lt", "Expr.lt", params={"other": 40}) # less than

# and node
g.add("and_node", "Expr.and_")

# filtering node
g.add("node_filter","LazyFrame.filter")

# preview(sink) node
g.add("node_preview","sink.preview")

g.connect("c_id","pred_gt",port="self")
g.connect("c_val","pred_lt",port="self")
g.connect("pred_gt","and_node",port="self")
g.connect("pred_lt","and_node",port="others")

g.connect("node_source","node_filter", port="self")

g.connect("and_node","node_filter",port="predicates")
g.connect("node_filter","node_preview")
```

```python
draw(g)
```

```python
run(g)
```
